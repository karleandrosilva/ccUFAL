receba uma lista de inteiros

contar o número de elementos negativos de um conjunto

```python
# contar quantidade de elementos negativos

def contar_negativo(lista):
    cont  = 0
    for numero in lista: # tem que comparar todos
        if numero < 0:
            cont += 1
    return cont

lista = [50, -1, 3, -5, 11, -54, 14, -5, -9] #lista de inteiros
print(f'Elementos negativos:', contar_negativo(lista))
```

identificar os valores de um conjunto que estão abaixo da média do conjunto

```python


def abaixoMedia(lista):
    for i in lista:
        i+=i
    return i

lista = [10, 20, 5, 10, 20, 30]
print(abaixoMedia(lista))
```
