```python
# BUSCA SEQUENCIAL COM FOR

def busca1(chave, lista):
    posicao = 0
    i = 0
    for nome in lista:
        if chave == nome:
            posicao = i
        i += 1

    print(f'numero de comparações: {i}')
    return posicao

lista = ['ana', 'bob', 'carlos', 'marta', 'daniel', 'pedro', 'beto', 'lucia'] # lista de nomes
chave = 'lucia'

print(f'posição: {busca1(chave, lista)}')
# 
```

```python
# COM WHILE

def busca2(chave, lista):
    i = 0
    posicao = 0
    while i < len(lista) and chave != lista[i]:
        i += 1
    if i == len(lista):
        posicao = None
    else:
        posicao = i
    print(f'numero de comparações: {i}')
    return posicao

lista = ['ana', 'bob', 'carlos', 'marta', 'daniel', 'pedro', 'beto', 'lucia']
chave = 'carlos'

print(f'posição: {busca2(chave, lista)}')
```

```python
# contar quantidade de elementos negativos

def contar_negativo(lista):
    cont  = 0
    for numero in lista: # tem que comparar todos
        if numero < 0:
            cont += 1
    return cont

lista = [50, -1, 3, -5, 11, -54, 14, -5, -9] #lista de inteiros
print(f'Elementos negativos:', contar_negativo(lista))
```
